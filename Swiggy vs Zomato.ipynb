# 📊 Univariate Analysis

Univariate analysis focuses on understanding the distribution and behavior of individual variables within the dataset. This section helps identify patterns, trends, skewness, and potential anomalies across financial, customer, operational, and competitive metrics.

```python
plt.figure(figsize=(10,5))

sns.histplot(
    df['swiggy_rating'],
    bins=20,
    kde=True,
    color='#FC8019',
    edgecolor='black'
)

plt.title("Distribution of Swiggy Ratings")

plt.show()
```

## ⚔️ Swiggy vs Zomato Rating Comparison

```python
plt.figure(figsize=(10,5))

sns.kdeplot(df['swiggy_rating'], label='Swiggy', color='#FC8019', fill=True)
sns.kdeplot(df['zomato_rating'], label='Zomato', color='#E23744', fill=True)

plt.title("Swiggy vs Zomato Rating Distribution")
plt.xlabel("Ratings")
plt.legend()

plt.show()
```

### Observation

- Both Swiggy and Zomato exhibit highly similar rating distributions, indicating comparable customer satisfaction patterns across platforms.
- The majority of restaurant ratings are concentrated around the 4.0 range, suggesting generally positive user experiences.
- Extremely low ratings are relatively uncommon, which may reflect consistent service quality or typical rating inflation behavior observed in food delivery platforms.

## 🚚 Impact of Delivery Time on Customer Ratings

```python
# Swiggy Correlation
swiggy_corr = df['swiggy_avg_delivery_time_minutes'].corr(df['swiggy_rating'])

# Zomato Correlation
zomato_corr = df['zomato_avg_delivery_time_minutes'].corr(df['zomato_rating'])

print("Swiggy Correlation:", round(swiggy_corr, 3))
print("Zomato Correlation:", round(zomato_corr, 3))
```

### Observation

The correlation between delivery time and customer ratings is extremely weak for both Swiggy (-0.014) and Zomato (0.023). This suggests that delivery speed alone does not significantly influence customer ratings within this dataset.

The findings indicate that customers may prioritize other factors such as food quality, pricing, packaging, and overall dining experience more heavily when rating restaurants on food delivery platforms.

## 💰 Impact of Discounts on Restaurant Profitability

```python
discount_profit = df.groupby('swiggy_discount_frequency_pct')['swiggy_estimated_net_profit_inr'].mean()

plt.figure(figsize=(10,5))

plt.plot(
    discount_profit.index,
    discount_profit.values,
    color='#FC8019',
    marker='o'
)

plt.xlabel("Discount Frequency (%)")
plt.ylabel("Average Net Profit (INR)")

plt.title("Average Profit by Discount Frequency - Swiggy")

plt.show()
```

```python
discount_profit_zomato = df.groupby('zomato_discount_frequency_pct')['zomato_estimated_net_profit_inr'].mean()

plt.figure(figsize=(10,5))

plt.plot(
    discount_profit_zomato.index,
    discount_profit_zomato.values,
    color='#E23744',
    marker='o'
)

plt.xlabel("Discount Frequency (%)")
plt.ylabel("Average Net Profit (INR)")

plt.title("Average Profit by Discount Frequency - Zomato")

plt.show()
```

### Observation

- Both Swiggy and Zomato exhibit fluctuating profit trends across different discount frequency levels, indicating that discounts alone do not consistently determine restaurant profitability.
- Moderate to higher discount frequencies occasionally correspond with increased average profits, suggesting that strategic discounting may help boost customer acquisition and order volumes.
- However, the presence of sharp rises and drops in profitability across similar discount levels highlights that additional factors such as operational efficiency, pricing strategy, cuisine type, and customer demand likely play a significant role in determining overall profit performance.
- Overall, the analysis suggests that discounts can support growth when applied strategically, but excessive or poorly optimized discounting may not always guarantee higher profitability.

## 📈 Relationship Between Orders and Revenue

```python
orders_revenue = df.groupby('swiggy_estimated_monthly_orders')['swiggy_estimated_monthly_revenue_inr'].mean()

plt.figure(figsize=(10,5))

plt.plot(
    orders_revenue.index,
    orders_revenue.values,
    color='#FC8019'
)

plt.xlabel("Estimated Monthly Orders")
plt.ylabel("Average Monthly Revenue (INR)")

plt.title("Orders vs Revenue - Swiggy")

plt.show()
```

```python
orders_revenue_zomato = df.groupby('zomato_estimated_monthly_orders')['zomato_estimated_monthly_revenue_inr'].mean()

plt.figure(figsize=(10,5))

plt.plot(
    orders_revenue_zomato.index,
    orders_revenue_zomato.values,
    color='#E23744'
)

plt.xlabel("Estimated Monthly Orders")
plt.ylabel("Average Monthly Revenue (INR)")

plt.title("Orders vs Revenue - Zomato")

plt.show()
```

```python
import warnings
warnings.filterwarnings('ignore')

bins = [75, 150, 225, 300, 375, 450]

labels = [
    '75-150',
    '151-225',
    '226-300',
    '301-375',
    '376-450'
]

df['swiggy_order_bins'] = pd.cut(
    df['swiggy_estimated_monthly_orders'],
    bins=bins,
    labels=labels
)

# Average Revenue by Order Bin
orders_revenue = df.groupby('swiggy_order_bins')['swiggy_estimated_monthly_revenue_inr'].mean()

# Plot
plt.figure(figsize=(10,5))

orders_revenue.plot(
    kind='bar',
    color='#FC8019'
)

plt.xlabel("Estimated Monthly Orders Range")
plt.ylabel("Average Revenue (INR)")

plt.title("Average Revenue by Order Range - Swiggy")

plt.xticks(rotation=0)

plt.show()
```

```python
# Creating Readable Order Bins

bins = [75, 150, 225, 300, 375, 450]

labels = [
    '75-150',
    '151-225',
    '226-300',
    '301-375',
    '376-450'
]

df['zomato_order_bins'] = pd.cut(
    df['zomato_estimated_monthly_orders'],
    bins=bins,
    labels=labels
)

# Average Revenue by Order Bin
orders_revenue_zomato = df.groupby('zomato_order_bins')['zomato_estimated_monthly_revenue_inr'].mean()

# Plot
plt.figure(figsize=(10,5))

orders_revenue_zomato.plot(
    kind='bar',
    color='#E23744'
)

plt.xlabel("Estimated Monthly Orders Range")
plt.ylabel("Average Revenue (INR)")

plt.title("Average Revenue by Order Range - Zomato")

plt.xticks(rotation=0)

plt.show()
```

### Observation

- Both Swiggy and Zomato demonstrate a clear positive relationship between order volume and average monthly revenue.
- Restaurants receiving higher monthly orders consistently generate significantly higher revenue across both platforms.
- The revenue growth pattern appears steady and scalable, indicating that increasing customer order volume is a major driver of business growth in food delivery platforms.
- Zomato shows slightly higher average revenue in the highest order range category, suggesting stronger revenue generation at larger order volumes within this dataset.
- Overall, the analysis highlights that order volume plays a critical role in determining restaurant revenue performance across both platforms.

## 🔍 Correlation Analysis of Business Metrics

```python
# Selecting Numerical Columns

corr_columns = [
    'swiggy_rating',
    'zomato_rating',
    'swiggy_avg_delivery_time_minutes',
    'zomato_avg_delivery_time_minutes',
    'swiggy_estimated_monthly_orders',
    'zomato_estimated_monthly_orders',
    'swiggy_estimated_monthly_revenue_inr',
    'zomato_estimated_monthly_revenue_inr',
    'swiggy_estimated_net_profit_inr',
    'zomato_estimated_net_profit_inr',
    'swiggy_discount_frequency_pct',
    'zomato_discount_frequency_pct'
]

# Correlation Matrix
corr_matrix = df[corr_columns].corr()

# Plot
plt.figure(figsize=(12,8))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap of Key Business Metrics")

plt.show()
```

### Observation

- Revenue and profitability exhibit extremely strong positive correlations across both Swiggy and Zomato, indicating that restaurants generating higher revenue also tend to achieve stronger financial performance.
- Monthly order volume demonstrates a moderate positive relationship with both revenue and profit, suggesting that higher customer demand contributes significantly to restaurant growth.
- Delivery time shows almost no meaningful correlation with customer ratings, reinforcing earlier findings that customer satisfaction may depend more heavily on food quality, pricing, and overall service experience.
- Discount frequency displays a notable positive relationship with ratings, indicating that promotional strategies and customer value perception may influence customer satisfaction levels.
- Strong correlations between Swiggy and Zomato operational metrics suggest that restaurant performance patterns remain relatively consistent across both platforms.

```python
# Filtering Restaurants

high_rating_low_profit_zomato = df[
    (df['zomato_rating'] >= 4.5) &
    (df['zomato_estimated_net_profit_inr'] < df['zomato_estimated_net_profit_inr'].median())
]

# Top Results
high_rating_low_profit_zomato[
    [
        'restaurant_name',
        'cuisines',
        'city',
        'zomato_rating',
        'zomato_estimated_net_profit_inr'
    ]
].head(10)
```

```python
# Correlation with Swiggy Profit

# Selecting Only Swiggy Metrics

swiggy_cols = [
    'swiggy_rating',
    'swiggy_total_reviews',
    'swiggy_delivery_fee_inr',
    'swiggy_avg_delivery_time_minutes',
    'swiggy_platform_commission_pct',
    'swiggy_discount_frequency_pct',
    'swiggy_estimated_monthly_orders',
    'swiggy_estimated_monthly_revenue_inr',
    'swiggy_estimated_net_profit_inr'
]

# Correlation with Profit
swiggy_profit_corr = df[swiggy_cols].corr()['swiggy_estimated_net_profit_inr'].sort_values(ascending=False)

# Removing self-correlation
swiggy_profit_corr = swiggy_profit_corr.drop('swiggy_estimated_net_profit_inr')

# Plot
plt.figure(figsize=(10,6))

sns.barplot(
    x=swiggy_profit_corr.values,
    y=swiggy_profit_corr.index,
    color='#FC8019'
)

plt.xlabel("Correlation with Net Profit")
plt.ylabel("Swiggy Business Metrics")

plt.title("Key Drivers of Swiggy Profitability")

plt.show()
```

```python
# Selecting Only Zomato Metrics

zomato_cols = [
    'zomato_rating',
    'zomato_total_reviews',
    'zomato_delivery_fee_inr',
    'zomato_avg_delivery_time_minutes',
    'zomato_platform_commission_pct',
    'zomato_discount_frequency_pct',
    'zomato_estimated_monthly_orders',
    'zomato_estimated_monthly_revenue_inr',
    'zomato_estimated_net_profit_inr'
]

# Correlation with Profit
zomato_profit_corr = df[zomato_cols].corr()['zomato_estimated_net_profit_inr'].sort_values(ascending=False)

# Removing self-correlation
zomato_profit_corr = zomato_profit_corr.drop('zomato_estimated_net_profit_inr')

# Plot
plt.figure(figsize=(10,6))

sns.barplot(
    x=zomato_profit_corr.values,
    y=zomato_profit_corr.index,
    color='#E23744'
)

plt.xlabel("Correlation with Net Profit")
plt.ylabel("Zomato Business Metrics")

plt.title("Key Drivers of Zomato Profitability")

plt.show()
```

### Conclusion

The profitability driver analysis highlights that revenue and monthly order volume are the strongest contributors to restaurant profitability across both Swiggy and Zomato. Restaurants generating higher customer demand and stronger revenue streams tend to achieve significantly better financial performance.

Customer ratings also demonstrate a positive relationship with profitability, indicating that customer satisfaction, food quality, and overall service experience contribute meaningfully to long-term business success.

Interestingly, discount frequency shows a moderate positive relationship with profitability, suggesting that strategic promotional campaigns may help improve customer acquisition and sales performance when managed effectively.

On the other hand, delivery time and total review count exhibit very weak or negligible relationships with profitability, indicating that operational speed and review quantity alone may not significantly influence financial outcomes within this dataset.

Overall, the analysis suggests that sustainable restaurant profitability is primarily driven by strong revenue generation, consistent customer demand, and positive customer experience rather than operational metrics alone.
